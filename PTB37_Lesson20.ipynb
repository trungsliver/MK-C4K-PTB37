## **Biến đổi dữ liệu**

**Hàm biến đổi dữ liệu cột Net Worth**

```python
def clean_networth(networth):
  # Xóa ký tự đầu và cuối: '$211 B' => '211 '
  return networth[1:-1].strip()
```

**DataFrame.apply(): thay đổi giá trị 1 cột bằng hàm tự định nghĩa**

```python
df["Net Worth"] = df["Net Worth"].apply(clean_networth)
```

**Chuyển đổi kiểu dữ liệu của cột**

```python
# pd.to_numeric(<cột>): chuyển string => float
df["Net Worth"] = pd.to_numeric(df["Net Worth"])
```

```python
# Biến đổi dữ liệu cột age: float -> int
df["Age"] = df["Age"].astype(int)
```

```python
df.info()
```

## **Nhóm và Sắp xếp**

Đề bài: "Đưa ra tổng tài sản lớn nhất của các tỉ phú nằm ở lĩnh vực nào"
<br>
=> Cần tính tổng lượng tài sản theo từng lĩnh vực

```python
# Tính tổng tài sản (Net Worth) theo lĩnh vực (Industry)
df_industry = df[['Industry', 'Net Worth']].groupby('Industry').sum()
df_industry
```

```python
# sắp xếp tăng dần
df_industry = df_industry.sort_values(by='Net Worth', ascending=False)
df_industry
```

Ví dụ 1: Tính tổng tài sản theo quốc gia
<br>
=> Top 5 quốc gia giàu nhất

```python
df_country = df[['Country | Territory', 'Net Worth']].groupby('Country | Territory').sum()
df_country = df_country.sort_values(by='Net Worth', ascending=False)
df_country.head(5)
```

Ví dụ 2: Số lượng tỷ phú mỗi quốc gia
<br>
=> Quốc gia có nhiều tỷ phú nhất

```python
amount_billionaires = df[['Country | Territory', 'Name']].groupby('Country | Territory').count()
amount_billionaires = amount_billionaires.sort_values(by='Name', ascending=False)
amount_billionaires.head(1)
```

dropna(): bỏ tất cả các hàng có giá trị bị khuyết: df.dropna()
<br>
Nếu muốn bỏ cột, thì kết hợp với axis=1: df.dropna(axis=1)

```python
# Loại bỏ 1 số cột không cần thiết: 'birthdate', 'birthplace'
df.drop(['birthdate', 'birthplace'], axis=1, inplace=True)
df
```

fillna(): thay thế các giá trị NaN thành giá trị mặc định

```python
# Thay thế giá trị NaN thành 0 trong cột goals
df.goals = df.goals.fillna(0)
df.info()
```

Kiểm tra dữ liệu lặp

```python
# Đếm số dòng trùng nhau
df.duplicated().sum()
```

```python
# Xóa các dòng trùng nhau
df.drop_duplicates()
```

Xử lý kiểu dữ liệu

```python
# Chuyển age: float => int
df.age = df.age.astype("int")
df.info()
```

Kiểm tra dữ liệu có điều kiện

```python
# Loại bỏ chiều cao, cân nặng âm
df = df[df["weight"] > 0]
df = df[df["height"] > 0]
df
```

Chuẩn hóa dữ liệu dạng text

```python
# Viết thường tất cả tên
df['name'] = df['name'].str.lower().str.strip()
df
```

**Bài tập**
- Top 3 cầu thủ có nhiều goals nhất
- Chiều cao trung bình của các cầu thủ
- Cầu thủ có cân nặng nhẹ nhất
- Cầu thủ lớn tuổi nhất và trẻ nhất
- Số lượng cầu thủ theo từng vị trí

```python
df = pd.read_csv('hockey_players_data.csv')

# Top 3 cầu thủ có nhiều bàn thắng nhất
top3_goals = df[['name', 'goals']].sort_values(by='goals', ascending=False)
print(top3_goals.head(3))

# Chiều cao trung bình
avg_height = df["height"].mean()
print("Chiều cao trung bình:", avg_height)

# Cầu thủ cân nặng nhẹ nhất
min_weight = df["weight"].min()
lightest_player = df[df["weight"] == min_weight]
print(lightest_player[["name", "weight"]])

# Lớn tuổi nhất
oldest = df[df["age"] == df["age"].max()]
print("Cầu thủ lớn tuổi nhất:")
print(oldest[["name", "age"]])

# Trẻ nhất
youngest = df[df["age"] == df["age"].min()]
print("\nCầu thủ trẻ nhất:")
print(youngest[["name", "age"]])

# Số lượng cầu thủ theo vị trí
position_count = df["position"].value_counts()
print(position_count)
```
