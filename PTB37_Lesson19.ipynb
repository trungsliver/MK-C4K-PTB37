# **DATA FRAME: Đọc - ghi file**

## **1. Đọc - ghi file trực tiếp**

```python
import pandas as pd
import numpy as np
```

```python
# Khởi tạo dữ liệu dạng dictionary
data = {
    'name': ['Hải Nam', 'Đức Trung', 'Đức Minh', 'Mỹ Linh'],
    'age': [14, 2, 13, 15],
    'gender': ['male', 'male', 'prefer not to say', 'female']
}
# Khởi tạo DataFrame
df = pd.DataFrame(data)
# Hiển thị dữ liệu từ DataFrame
df
```

## **Tạo và đọc dữ liệu**

```python
# Tạo dữ liệu CSV
df.to_csv('data.csv', index=False)

# Đọc dữ liệu CSV
df_csv = pd.read_csv('data.csv')
df_csv
```

```python
# tải file về máy
from google.colab import files
files.download('data.csv')
```

```python
# Tạo dữ liệu xlsx (file excel)
df = pd.DataFrame(data)
df.to_excel('data.xlsx', index=False)

# Đọc dữ liệu xlsx
df_xlsx = pd.read_excel('data.xlsx')
df_xlsx
```

```python
# Tạo dữ liệu dạng json
df.to_json('data.json')

# Đọc dữ liệu dạng json
df_json = pd.read_json('data.json')
df_json
```

## **2. Truy cập theo hàng**

slice[start:end] => chạy từ start đến end-1

```python
# Hiện từ hàng 5 đến 8
animals[4:9]
```

```python
# Hiện từ đầu: [:end]
animals[:4]
```

## **3. Truy cập theo cả hàng và cột**

# **Tổng hợp 1 số lệnh hay dùng với DataFrame**

```python
# Khởi tạo dữ liệu dạng dictionary
data = {
    'name': ['Hải Nam', 'Đức Trung', 'Đức Minh', 'Mỹ Linh'],
    'age': [14, 2, 13, 15],
    'gender': ['male', 'male', 'prefer not to say', 'female']
}
# Khởi tạo DataFrame
df = pd.DataFrame(data)
# Hiển thị dữ liệu từ DataFrame
df
```

```python
# Xem kiểu dữ liệu của từng cột
df.dtypes
```

```python
# Xem số hàng và cột
df.shape
```

```python
# Tóm tắt thông tin của DataFrame: số dòng, số cột, số giá trị có dữ liệu, kiểu dữ liệu
df.info()
```

```python
# Thống kê các cột dạng số
    # count: số dữ liệu (bản ghi)
    # mean: trung bình
    # std: độ lệch chuẩn
    # min, max: giá trị lớn/nhỏ nhất
df.describe()
```

```python
# Hiển thị các dòng đầu: df.head(n)
df.head(2)
```

```python
# Hiển thị các dòng cuối: df.tail(n)
df.tail(3)
```

```python
# Đổi tên trục: rename_axis()
    # index (axis=0) → trục hàng
    # columns (axis=1) → trục cột
df = df.rename_axis(index=None) # Xóa tên cột index
df
```

```python
# Đặt 1 cột làm index: set_index()
df = df.set_index('name')
df
```

```python
# Đưa index trở lại thành cột: reset_index()
df.reset_index()
```

```python
# Đổi tên cột hoặc index: df.rename(columns={"old_name":"new_name"})
df = df.rename(columns={"age":"tuổi"})
df
```
